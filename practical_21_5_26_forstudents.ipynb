# 3. Check Space Weather Data

- **Solar flare level**

The X-ray flux is classified based on the GOES XRS-B flux value in W/m²:

 **X-class flare:** flux ≥ 1 × 10⁻⁴ W/m²  
 **M-class flare:** flux ≥ 1 × 10⁻⁵ W/m²  
 **C-class flare:** flux ≥ 1 × 10⁻⁶ W/m²  
 **B-class flare:** flux ≥ 1 × 10⁻⁷ W/m²  
 **A-class or very weak:** flux < 1 × 10⁻⁷ W/m²
  
- **Kp index**

The Kp index describes the level of geomagnetic activity.  

Kp ranges from 0 to 9, where higher values mean stronger geomagnetic disturbance.

  **Kp 0–2:** Quiet geomagnetic conditions  
  **Kp 3:** Unsettled conditions  
  **Kp 4:** Active conditions  
  **Kp 5 (G1):** Minor geomagnetic storm  
  **Kp 6 (G2):** Moderate geomagnetic storm  
  **Kp 7 (G3):** Strong geomagnetic storm  
  **Kp 8(G4):** Severe geomagnetic storm  
  **Kp 9 (G5):** Extreme geomagnetic storm
