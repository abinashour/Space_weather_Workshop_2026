# 1. Introduction

- What is space weather?

  Changing condition in space environment caused by the sun.
---
- Why do we monitor the Sun ?

  We monitor the sun because solar activity can affect earth and space technology.
  
  Solar flares and CMEs can disturb satellites, GPS signal, Spacecraft operations and astronauts also they can cause ionosphere disturbance. Strong solar activity can cause aurora lights and geomagnetic storms, which affect power grids and pipelines.
---
- What are sunspots, solar flares, CMEs, Kp index?

  Sunspots are dark spots on the sun's surface which they are much cooler than the surrounding area. Also, they are linked to solar activity occurrence.

  Solar flares are sudden burst of electromagnetic radiation including X-ray and Ultra ultraviolet radiation from the sun.

  CMEs are explosion of plasma and magnetic fields from the sun's corona to the space.

  Kp index is a number from 0 to 9 tell us how Earth's magnetic field is  disturbed because of solar activity.

# 3. Check Space Weather Data

- **Solar flare level**

The X-ray flux is classified based on the GOES XRS-B flux value in W/m²:

 **X-class flare:** flux ≥ $1\times 10^-4 \ \mathrm{W/m^2}$

 **M-class flare:** flux ≥ $1\times 10^-5 \ \mathrm{W/m^2}$

 **C-class flare:** flux ≥ $1\times 10^-6 \ \mathrm{W/m^2}$

 **B-class flare:** flux ≥ $1\times 10^-7 \ \mathrm{W/m^2}$

 **A-class flare:** flux ≥ $1\times 10^-8 \ \mathrm{W/m^2}$

 **Very low X-ray background:** flux < $1\times 10^-8 \ \mathrm{W/m^2}$
  
- **Kp index**

The Kp index describes the level of geomagnetic activity.  

Kp ranges from 0 to 9, where higher values mean stronger geomagnetic disturbance.

  **Kp 0–2:** Quiet geomagnetic conditions  
  **Kp 3:** Unsettled conditions  
  **Kp 4:** Active conditions  
  **Kp 5 (G1):** Minor geomagnetic storm  
  **Kp 6 (G2):** Moderate geomagnetic storm  
  **Kp 7 (G3):** Strong geomagnetic storm  
  **Kp 8(G4):** Severe geomagnetic storm  
  **Kp 9 (G5):** Extreme geomagnetic storm  

**Read GOES X-ray flux CSV file**

```python

flare_2020 = pd.read_csv(r"/content/Space_weather_Workshop_2026/required data and images/27-28_01_2020/flare_27_Jan_2020.csv") # x-ray flux file
flare_2026 = pd.read_csv(r"/content/Space_weather_Workshop_2026/required data and images/18-19_01_2026/flare_18_Jan_2026.csv") # x-ray flux file
```

```python
#Read what variabels contained in the CSV file and display the first two rows

flare_2020.head(2)
#flare_2026.head(2)
```

```python
# This function classify Xray flux

def classify_xray_flux(flux):
    if flux >= 1e-4:
        return "X-class flare"
    elif flux >= 1e-5:
        return "M-class flare"
    elif flux >= 1e-6:
        return "C-class flare"
    elif flux >= 1e-7:
        return "B-class flare"
    elif flux >= 1e-8:
        return "A-class flare"
    else:
        return "Very low X-ray background"
```

```python
# Get peak XRS-B flux value

peakxrs_2020 = flare_2020.loc[flare_2020["xrsb_flux"].idxmax()]
peakxrs_2026 = flare_2026.loc[flare_2026["xrsb_flux"].idxmax()]
```

```python
# Make column readable

flare_2020["time_tag"] = pd.to_datetime(flare_2020["time_tag"], errors="coerce")
flare_2026["time_tag"] = pd.to_datetime(flare_2026["time_tag"], errors="coerce")
flare_2020["xrsb_flux"] = pd.to_numeric(flare_2020["xrsb_flux"], errors='coerce')
flare_2026["xrsb_flux"] = pd.to_numeric(flare_2026["xrsb_flux"], errors='coerce')
```

**Plot time vs. xrsb flux**

```python
plt.figure(figsize=(10, 5))

plt.plot(flare_2020["time_tag"], flare_2020["xrsb_flux"], label="XRS-B Flux, 1–8 Å")

plt.yscale("log")
plt.xlabel("Time UTC")
plt.ylabel("X-ray Flux (W/m²)")
plt.title("GOES-18 X-ray Flux: 27 January 2020")
#plt.grid(True, which="both", alpha=0.3)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()

print("Maximum X-ray Flux:")
print("Time:", peakxrs_2020["time_tag"], "UTC")
print("Flux:", f"{peakxrs_2020['xrsb_flux']:.2e}", "W/m²")
print("Maximum flux classification:",classify_xray_flux(peakxrs_2020["xrsb_flux"]))
```

```python
plt.figure(figsize=(10, 5))

plt.plot(flare_2026["time_tag"], flare_2026["xrsb_flux"], label="XRS-B Flux, 1–8 Å")

plt.yscale("log")
plt.xlabel("Time UTC")
plt.ylabel("X-ray Flux (W/m²)")
plt.title("GOES-18 X-ray Flux: 18 January 2026")
#plt.grid(True, which="both", alpha=0.3)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()

print("Maximum X-ray Flux:")
print("Time:", peakxrs_2026["time_tag"], "UTC")
print("Flux:", f"{peakxrs_2026['xrsb_flux']:.2e}", "W/m²")
print("Maximum flux classification:",classify_xray_flux(peakxrs_2026["xrsb_flux"]))
```

**Read Kp index CSV file**

```python

kp_2020 = pd.read_csv(r"/content/Space_weather_Workshop_2026/required data and images/27-28_01_2020/Kp_27_28_Jan_2020.csv") # Kp index file
kp_2026 = pd.read_csv(r"/content/Space_weather_Workshop_2026/required data and images/18-19_01_2026/Kp_18_19_Jan_2026.csv") # Kp index file
```

```python
#Read what variabels contained in the csv file and display the first two rows

kp_2020.head(2)
#kp_2026.head(2)
```

```python
# Get peak Kp value

peakkp_2020 = kp_2020.loc[kp_2020["Kp"].idxmax()]
peakkp_2026 = kp_2026.loc[kp_2026["Kp"].idxmax()]
```

```python
# Make column readable

kp_2020 ["time"] = pd.to_datetime(kp_2020 ["time"], errors="coerce")
kp_2026 ["time"] = pd.to_datetime(kp_2026 ["time"], errors="coerce")
kp_2020 ["Kp"] = pd.to_numeric(kp_2020 ["Kp"], errors='coerce')
kp_2026 ["Kp"] = pd.to_numeric(kp_2026 ["Kp"], errors='coerce')
```

```python
# This function gives each Kp value a color similar to the NOAA/SWPC style:
# low Kp = green, storm levels = yellow/gold/orange/red/darkred.

def kp_color(kp):
    if kp < 5:
        return "green"       # quiet to active
    elif kp < 6:
        return "yellow"      # G1 minor storm
    elif kp < 7:
        return "gold"      # G2 moderate storm
    elif kp < 8:
        return "orange"         # G3 strong storm
    elif kp < 9:
        return "red"      # G4 severe storm
    else:
        return "darkred"       # G5 extreme storm
```

**Plot time vs. Kp index histogram**

```python
colors = kp_2020["Kp"].apply(kp_color)

plt.figure(figsize=(10, 4))

plt.bar(kp_2020["time"], kp_2020["Kp"], color=colors, width=0.1, label="Kp index")


plt.xlabel("Time")
plt.ylabel("Kp Index")
plt.title("Kp Index: 27–28 Jan 2020")
plt.ylim(0, 9)
plt.grid(True)
plt.legend()
plt.tight_layout()

plt.show()
print("Maximum Kp:")
print("Time:", peakkp_2020["time"], "UTC")
print("Kp:", peakkp_2020["Kp"])
```

```python
colors = kp_2026["Kp"].apply(kp_color)
plt.figure(figsize=(10, 4))

plt.bar(kp_2026["time"], kp_2026["Kp"], color=colors, width=0.1, label="Kp index")


plt.xlabel("Time")
plt.ylabel("Kp Index")
plt.title("Kp Index: 18–19 Jan 2026")
plt.ylim(0, 9)
plt.grid(True)
plt.legend()
plt.tight_layout()

plt.show()
print("Maximum Kp:")
print("Time:", peakkp_2026["time"], "UTC")
print("Kp:", peakkp_2026["Kp"])
```

# 4. Classify Space Weather

- Quiet: Few sunspots, no strong flare, no CME, low Kp
- Stormy: Complex sunspot groups, M/X flare, Earth-directed CME, high Kp
